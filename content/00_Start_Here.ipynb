# JupyterLite + World Monitor prototype

This prototype uses **World Monitor's public sandbox** from a browser-only JupyterLite environment. The sandbox is deterministic, schema-valid sample data: it is ideal for learning the API shape, building parsers, and prototyping analysis, but it is **not live intelligence**.

The current World Monitor Free dashboard does not include REST API or MCP quota. Production API access requires an eligible paid plan, so this public GitHub Pages deployment intentionally does not store an API key.

## What you can do here

1. Discover the sandbox operations dynamically.
2. Load representative country-risk, conflict, market, chokepoint, forecast, and resilience responses.
3. Build analysis code against the same response envelopes used by production.
4. Later swap the data-access layer for an authenticated backend without rewriting most notebook analysis.
